```python
import numpy as np
import soundfile as sf
from IPython.display import Audio, display

def text_to_audiobook(file_path, output_filename="audiobook.wav", voice="pm_alex", ):
    # 1. Ler o arquivo de texto
    with open(file_path, 'r', encoding='utf-8') as f:
        full_text = f.read()

    # 2. Dividir o texto em partes menores (por pontos finais)
    # O Kokoro performa melhor com sentenças de até 200-300 caracteres
    sentences = full_text.replace('\n', ' ').split('. ')

    all_audio = []
    print(f"Iniciando processamento de {len(sentences)} sentenças...")

    for i, sentence in enumerate(sentences):
        if len(sentence.strip()) < 2: continue # Pula linhas vazias

        # Gerar áudio para a sentença atual
        samples, sample_rate = kokoro.create(sentence + ".", voice=voice, speed=0.85, lang="pt-br")
        all_audio.append(samples)

        if i % 5 == 0:
            print(f"Progresso: {i}/{len(sentences)}")

    # 3. Concatenar todos os pedaços
    final_audio = np.concatenate(all_audio)

    # 4. Salvar o arquivo final
    sf.write(output_filename, final_audio, sample_rate)
    print(f"✅ Audiobook gerado com sucesso: {output_filename}")
    return output_filename
```

```python
import numpy as np
import soundfile as sf
from pydub import AudioSegment
from IPython.display import Audio, display

def text_to_mp3(file_path, output_mp3="audiobook.mp3", voice="pf_dora"):
    # 1. Ler o texto
    with open(file_path, 'r', encoding='utf-8') as f:
        full_text = f.read()

    sentences = full_text.replace('\n', ' ').split('. ')
    all_audio = []

    print(f"🎙️ Processando {len(sentences)} sentenças...")

    for i, sentence in enumerate(sentences):
        if len(sentence.strip()) < 2: continue

        # Gerar áudio
        samples, sample_rate = kokoro.create(sentence + ".", voice=voice, speed=0.66, lang="pt-br")
        all_audio.append(samples)

        if i % 10 == 0:
            print(f"Progresso: {i}/{len(sentences)}")

    # 2. Concatenar e salvar em WAV temporário
    final_audio = np.concatenate(all_audio)
    temp_wav = "temp_output.wav"
    sf.write(temp_wav, final_audio, sample_rate)

    # 3. Converter WAV para MP3 usando Pydub
    print("📦 Convertendo para MP3...")
    audio_segment = AudioSegment.from_wav(temp_wav)
    audio_segment.export(output_mp3, format="mp3", bitrate="192k")

    # Limpeza: remover o arquivo WAV temporário
    import os
    os.remove(temp_wav)

    print(f"✅ Concluído! Arquivo disponível: {output_mp3}")
    return output_mp3

# Execução
arquivo_final = text_to_mp3("meu_livro.txt")#, voice="pm_santa")
display(Audio(arquivo_final))
```
